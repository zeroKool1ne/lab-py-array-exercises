
8. **Check for Duplicates in Array**
    - Write a Python program to find out if a given array of integers contains any duplicate elements.

    **Example:**
    ```python
    # Sample Input
    arr = [1,2,3,4]

    # Sample Output
    Contains duplicates: False

    arr = [1,2,2]

    Contains duplicates: True
    ```


````python
def has_duplicates(a):
    """A set drops duplicates, so a shorter set means at least one duplicate."""
    return len(set(a)) != len(a)

print(has_duplicates(array("i", [1, 2, 3, 4])))  # False
print(has_duplicates(array("i", [1, 2, 3, 2])))  # True
````

    - Write a Python program to count how many even and odd numbers are present in an integer array.

    **Example:**
    ```python
    # Sample Input
    arr = [10 ,21 ,32 ,43 ,54 ,65 ,76 ,87]

    # Sample Output
    Even count:4

    ```


````python
arr = array("i", [1, 2, 3, 4, 5, 6, 7])
even = sum(1 for x in arr if x % 2 == 0)
odd = len(arr) - even
print("Even:", even, "Odd:", odd)
````

1. **Rotate Array Elements**
    - Write a Python program to rotate elements of an integer array by `k` positions.

    **Example:**
    ```python
    # Sample Input
    arr = [10 ,20 ,30 ,40 ,50]
    k =2

    # Sample Output
    Rotated Array: [30 ,40 ,50 ,10 ,20]
    ```


````python
def rotate(a, k):
    """Rotate to the right by k positions."""
    k = k % len(a)  # k can be larger than the length
    return a[-k:] + a[:-k] if k else a[:]

arr = [1, 2, 3, 4, 5]
print("Original:", arr)
print("Rotated by 2:", rotate(arr, 2))
````


2. **Merge Two Sorted Arrays**
     - Write a Python program to merge two sorted arrays into one sorted array.

     **Example:**
     ```python
     # Sample Input
     arr1 = [1 ,3 ,5]
     arr2 = [2 ,4 ,6]

     # Sample Output
     Merged Sorted Array: [1 ,2 ,3 ,4 ,5 ,6]
     ```



````python
def merge_sorted(a, b):
    """Two-pointer merge, runs in O(n + m)."""
    i = j = 0
    merged = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            merged.append(a[i]); i += 1
        else:
            merged.append(b[j]); j += 1
    merged.extend(a[i:])
    merged.extend(b[j:])
    return merged

print(merge_sorted([1, 3, 5, 7], [2, 4, 6, 8, 10]))
````
